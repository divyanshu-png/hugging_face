```python
#@title Text generation
#text generation using GPT2 model from Hugging face Library

#use a pipline as a high-level helper
from transformers import pipeline

pipe = pipeline("text-generation", model="openai-community/gpt2")
```

```python
prompt="What is machine learninn? "
output= pipe(prompt)
```

```python
output[0]["generated_text"]
```
