Tokenization is the essential process of breaking down human-readable text into smaller units called tokens. These tokens can be words, parts of words, or even individual characters.

What Tokenization Signifies:

1. Segmentation: The model segments complex words (like "unbelievable" or "homoscedasticity") into smaller, manageable chunks. This allows the model to handle new or highly complex vocabulary effectively.
2. Mapping to IDs: Once text is split, each token is mapped to a specific numerical ID . Computers cannot process raw text; they require these numerical representations to perform mathematical operations
3. Foundation for Processing: These IDs serve as the bridge between human language and machine learning architectures, such as the Transformer.

Why it is Useful:
1. Data Pre-processing: Tokenization transforms unstructured text into structured numerical data, which is a mandatory step before any prompt can be passed to a language model like GPT-2.
2. Handling Complexity: By using sub-word tokenization, models can understand and process words they have never seen before by breaking them down into known, smaller segments.
3. Efficient Representation: It enables the model to convert tokens into embedding vectors, which capture the semantic and contextual meaning of the text, allowing the AI to "understand" language patterns and relationships.

```python
sentence2 = 'i learn machine learning to enhance the performace'
generated_text = sentence2

while True:
    input_ids_sentence = tokenizer(generated_text, return_tensors='pt').input_ids
    outputs = GPT2(input_ids_sentence) # Get the full outputs from the model
    next_token_logits = outputs.logits[0, -1] # Get logits for the last token
    next_token_id = next_token_logits.argmax() # Get the ID of the most probable next token
    predicted_word = tokenizer.decode(next_token_id) # Decode the token ID to a word

    if predicted_word == '.':
        generated_text += predicted_word
        break # Exit loop if a period is predicted

    generated_text += predicted_word # Append the predicted word to the sentence

    # Optional: add a safety break for extremely long sentences
    if len(generated_text.split()) > 50: # Limit to 50 words to prevent infinite loops
        break

print(generated_text)
```

```python
#sampling stratergies

def greedy_decode(logits):
    """Returns the token index with maximum probability"""
    return torch.argmax(logits, dim=-1) #we don't need to specify the dim, it automatically can choose the last dimension that is the -1 index

greedy_decode(output)
```

```python
tokenizer.decode(greedy_decode(output))
```

```python
import torch.nn.functional as F
```

```python
def top_k_sampling(logits, k=50):
    """Keeps only Top k logits, normalize them into probablity using the softmax function. Then sample one token from the filtered distribution."""
    values, indices= torch.topk(logits, k )
    probs = F.softmax(values, dim=-1 )
    sampled= torch.multinomial(probs, 1)
    return indices[sampled]

top_k_sampling(output)
```

```python
tokenizer.decode(top_k_sampling(output))
```
